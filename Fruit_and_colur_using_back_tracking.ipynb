```python
# Fruit and Color using Backtracking

fruits = ["apple", "banana", "orange"]
colors = ["red", "yellow", "orange"]

solution = {}

def backtrack(index):
    # If all fruits are assigned a color
    if index == len(fruits):
        return True

    fruit = fruits[index]

    for color in colors:

        # Check whether this color is already used
        if color not in solution.values():

            # Assign color
            solution[fruit] = color

            # Try the next fruit
            if backtrack(index + 1):
                return True

            # If it doesn't work, undo the assignment
            del solution[fruit]

    return False


# Start backtracking
if backtrack(0):
    print("Fruit and Color:")
    for fruit, color in solution.items():
        print(fruit, "->", color)
else:
    print("No solution found.")
```
