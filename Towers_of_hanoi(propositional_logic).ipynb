```python
# Propositional Logic representation
# A1 means: Disk 1 is on Rod A
# B1 means: Disk 1 is on Rod B
# C1 means: Disk 1 is on Rod C

state = {
    "A1": True,
    "A2": True,
    "A3": True,

    "B1": False,
    "B2": False,
    "B3": False,

    "C1": False,
    "C2": False,
    "C3": False
}

def move_disk(disk, source, destination):

    source_symbol = source + str(disk)
    destination_symbol = destination + str(disk)

    if state[source_symbol]:

        # Check whether destination is empty
        destination_empty = True

        for d in range(1, 4):
            if state[destination + str(d)]:
                destination_empty = False

        if destination_empty:

            state[source_symbol] = False
            state[destination_symbol] = True

            print(f"Move disk {disk}: {source} -> {destination}")

        else:
            print("Move is not valid")

# Example
move_disk(1, "A", "B")

print(state)
```
