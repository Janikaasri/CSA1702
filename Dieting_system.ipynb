```python
#Dieting System Based on Disease

diets = {
    "diabetes": [
        "Vegetables",
        "Whole grains",
        "Nuts",
        "Low-sugar fruits"
    ],

    "hypertension": [
        "Fruits",
        "Vegetables",
        "Whole grains",
        "Low-salt foods"
    ],

    "anemia": [
        "Spinach",
        "Beans",
        "Iron-rich foods",
        "Vitamin C rich fruits"
    ],

    "obesity": [
        "Vegetables",
        "Salads",
        "Whole grains",
        "Low-calorie foods"
    ]
}

# Get disease from user
disease = input("Enter the disease: ").lower()

# Check disease and suggest diet
if disease in diets:
    print("\nRecommended diet for", disease, ":")

    for food in diets[disease]:
        print("-", food)

else:
    print("Disease not found in the system.")
```
