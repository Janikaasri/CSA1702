```python
from itertools import permutations

students = ["A", "B", "C"]
seats = [1, 2, 3]

print("Valid Seat Assignments:\n")

for assignment in permutations(seats):
    A, B, C = assignment

    # Constraints
    if A != 2 and B > C:
        print(f"A -> Seat {A}, B -> Seat {B}, C -> Seat {C}")
```
