```python
# Tower of Hanoi with Disk Sizes

# -----------------------------
# 1. Assign size to each disk
# -----------------------------
size = {
    1: 1,   # Disk 1 = smallest
    2: 2,   # Disk 2
    3: 3    # Disk 3 = largest
}

# -----------------------------
# 2. Store disks on each rod
# -----------------------------
rods = {
    "A": [3, 2, 1],   # Source
    "B": [],          # Auxiliary
    "C": []           # Destination
}


# -----------------------------
# 3. Check whether a move is valid
# -----------------------------
def valid_move(source, destination):

    # Source must contain a disk
    if len(rods[source]) == 0:
        return False

    moving_disk = rods[source][-1]

    # Destination is empty
    if len(rods[destination]) == 0:
        return True

    top_disk = rods[destination][-1]

    # Smaller disk can be placed on larger disk
    if size[moving_disk] < size[top_disk]:
        return True

    return False


# -----------------------------
# 4. Perform a disk movement
# -----------------------------
def move_disk(source, destination):

    if valid_move(source, destination):

        disk = rods[source].pop()
        rods[destination].append(disk)

        print(
            "Move Disk", disk,
            "from", source,
            "to", destination
        )

        print("A:", rods["A"])
        print("B:", rods["B"])
        print("C:", rods["C"])
        print()

    else:
        print(
            "Invalid move:",
            source, "->", destination
        )


# -----------------------------
# 5. Tower of Hanoi Algorithm
# -----------------------------
def hanoi(n, source, auxiliary, destination):

    # Base case
    if n == 1:
        move_disk(source, destination)
        return

    # Move smaller disks to auxiliary rod
    hanoi(
        n - 1,
        source,
        destination,
        auxiliary
    )

    # Move largest remaining disk
    move_disk(source, destination)

    # Move smaller disks to destination
    hanoi(
        n - 1,
        auxiliary,
        source,
        destination
    )


# -----------------------------
# 6. Start the program
# -----------------------------
print("Initial State:")
print("A:", rods["A"])
print("B:", rods["B"])
print("C:", rods["C"])
print()

hanoi(3, "A", "B", "C")

print("Final State:")
print("A:", rods["A"])
print("B:", rods["B"])
print("C:", rods["C"])
```
