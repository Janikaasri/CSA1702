```python
# Family Tree

# Parent relationships
parent = {
    "Ravi": ["Arun", "Priya"],
    "Meena": ["Arun", "Priya"],
    "Arun": ["Kiran"],
    "Priya": ["Kiran"]
}

# Gender information
male = ["Ravi", "Arun", "Kiran"]
female = ["Meena", "Priya"]


# Check whether X is a parent of Y
def is_parent(x, y):
    return y in parent.get(x, [])


# Check whether X is a child of Y
def is_child(x, y):
    return is_parent(y, x)


# Check whether X is a father of Y
def is_father(x, y):
    return is_parent(x, y) and x in male


# Check whether X is a mother of Y
def is_mother(x, y):
    return is_parent(x, y) and x in female


# Check whether X and Y are siblings
def is_sibling(x, y):
    for p in parent:
        if x in parent[p] and y in parent[p]:
            return True
    return False


# Display relationships
print("Is Ravi the father of Arun?")
print(is_father("Ravi", "Arun"))

print("\nIs Meena the mother of Priya?")
print(is_mother("Meena", "Priya"))

print("\nIs Kiran the child of Arun?")
print(is_child("Kiran", "Arun"))

print("\nAre Arun and Priya siblings?")
print(is_sibling("Arun", "Priya"))
```
